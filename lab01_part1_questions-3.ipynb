### Question 2

We can also use other distance metrics such as L1 distance.
For pixel values $p_{ij}^{(k)}$ at location $(i,j)$ of some image $I_k$,

the mean $\mu$ across all pixels over all images is $$\mu=\frac{1}{nhw}\sum_{k=1}^n\sum_{i=1}^{h}\sum_{j=1}^{w}p_{ij}^{(k)}$$
And the pixel-wise mean $\mu_{ij}$ across all images is
$$\mu_{ij}=\frac{1}{n}\sum_{k=1}^np_{ij}^{(k)}.$$
The general standard deviation $\sigma$ and pixel-wise standard deviation $\sigma_{ij}$ is defined similarly.

Which of the following preprocessing steps will not change the performance of a Nearest Neighbor classifier that uses L1 distance? Select all that apply.
1. Subtracting the mean $\mu$ ($\tilde{p}_{ij}^{(k)}=p_{ij}^{(k)}-\mu$.)
2. Subtracting the per pixel mean $\mu_{ij}$  ($\tilde{p}_{ij}^{(k)}=p_{ij}^{(k)}-\mu_{ij}$.)
3. Subtracting the mean $\mu$ and dividing by the standard deviation $\sigma$.
4. Subtracting the pixel-wise mean $\mu_{ij}$ and dividing by the pixel-wise standard deviation $\sigma_{ij}$.
5. Rotating the coordinate axes of the data.

*Your Answer:* *1,2,3*


*Your Explanation:*

1. Тк $\mu$ - общее среднее и не зависит от пикселя, то при вычитании из каждой "координаты" просто сократится, на примере расстояния между изображениями А (тест) и Б (обучающее):$$
d_{L1}(A,B) = \sum_{i=1}^{h}\sum_{j=1}^{w}\left|p_{ij}^{A} - \mu - (p_{ij}^{B} - \mu )\right| = \sum_{i=1}^{h}\sum_{j=1}^{w}\left|p_{ij}^{A}-p_{ij}^{B}\right|.$$
2. Тк расстояние считается между одними и теми же пикселями, то из соответствующих пикселей обоих изображений вычитается одно и то же $\mu_{ij}$, поэтому оно сокращается аналогично предыдущему вопросу: $$
d_{L1}(A,B) = \sum_{i=1}^{h}\sum_{j=1}^{w}\left|p_{ij}^{A} - \mu_{ij} - (p_{ij}^{B} - \mu_{ij} )\right| = \sum_{i=1}^{h}\sum_{j=1}^{w}\left|p_{ij}^{A}-p_{ij}^{B}\right|.$$
3. Рассмотрим $\sigma$: новое расстояние $$
 d_{L1}(A,B)= \sum_{i=1}^{h}\sum_{j=1}^{w}\left|\frac{1}{\sigma}(p_{ij}^{A} - \mu) - \frac{1}{\sigma}(p_{ij}^{B} - \mu )\right| = \sum_{i=1}^{h}\sum_{j=1}^{w}\left|\frac{1}{\sigma}(p_{ij}^{A} - p_{ij}^{B})\right| = |\frac{1}{\sigma}| \sum_{i=1}^{h}\sum_{j=1}^{w}\left|(p_{ij}^{A} - p_{ij}^{B})\right|$$
Расстояние изменилось, но тк $\sigma$ - единое число для всех пикселей и изображений, то все расстояния умножатся на один и тот же коэффициент (вес у всех пикселей останется одинаковым), поэтому дальнейшая логика kNN не изменится и результат останется прежним
4. Рассмотрим $\sigma_{ij}$: $$
 d_{L1}(A,B)= \sum_{i=1}^{h}\sum_{j=1}^{w}\left|\frac{1}{\sigma_{ij}}(p_{ij}^{A} - \mu_{ij}) - \frac{1}{\sigma_{ij}}(p_{ij}^{B} - \mu_{ij} )\right| = \sum_{i=1}^{h}\sum_{j=1}^{w}\left|\frac{1}{\sigma_{ij}}(p_{ij}^{A} - p_{ij}^{B})\right| $$
 Раньше у пикселей был единый вес = 1, а теперь он разный и зависит от пикселя ($\frac{1}{\sigma_{ij}}$), поэтому расстояния после таких преобразований изменятся не одинаково и порядок может измениться, как и результат kNN
5. L1-норма не инвариантна относительно поворота системы координат. Тк точки с одинаковой L1-нормой образуют ромб, то при повороте системы координат расстояние до точек ромба не сохраняется, тк ромб не инвариантен относительно поворотов (в отличии от окружности)




## Question 3

Which of the following statements about $k$-Nearest Neighbor ($k$-NN) are true in a classification setting, and for all $k$? Select all that apply.
1. The decision boundary (hyperplane between classes in feature space) of the k-NN classifier is linear.
2. The training error of a 1-NN will always be lower than that of 5-NN.
3. The test error of a 1-NN will always be lower than that of a 5-NN.
4. The time needed to classify a test example with the k-NN classifier grows with the size of the training set.
5. None of the above.

*Your Answer:* $2, 4$


*Your Explanation:*

1. Ответ: нет. В kNN локально сравниваются соседи, поэтому граница может состоять из нескольких отдельных участков и составлять сложные формы. Наиболее простой пример:
- Класс А: (-1, -1) и (1, 1)
- Класс Б: (-1,  1) и (1, -1)

Для определения класса необходимо провести две прямые вдоль осей координат, разбив плоскость на 4 части. И в этом случае провести одну прямую для однозначного разделения точек разных классов невозможно.

2. Ответ: да. Тк для 1-NN при тестировании на обучающей выборке ближайший обучающий объект для тестового - он сам. Поэтому для 1-NN ошибка будет равна 0. При 5-NN добавляются соседние объекты, из-за которых может возрастать ошибка.
3. Ответ: нет. Особенно, если выборка шумная. В таком случае больше вероятность, что класс ближайшего соседа может оказаться неправильным. При 5-NN голосование нескольких соседей может уменьшить влияние отдельных шумных объектов и класс может определяться более правильно. Тогда 5-NN будет иметь меньшую ошибку.
4. Ответ: да. Тк в kNN вычисляются расстояния до всех обучающих объектов, и при возрастании их количества время вычисления тоже вырастет
